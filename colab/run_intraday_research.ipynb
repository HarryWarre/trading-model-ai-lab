```python
import os, subprocess, pathlib
REPO='HarryWarre/trading-model-ai-lab'
WORK='/content/trading-model-ai-lab'
try:
    if not pathlib.Path(WORK).exists():
        subprocess.run(['git','clone',f'https://github.com/{REPO}.git',WORK], check=True, capture_output=True, text=True)
    else:
        subprocess.run(['git','-C',WORK,'pull','--ff-only'], check=True, capture_output=True, text=True)
    os.chdir(WORK)
    subprocess.run(['pip','install','-q','pandas','numpy','scikit-learn'], check=True)
    print("Setup completed successfully.")
except subprocess.CalledProcessError as e:
    print(f"Error occurred: {e.stderr}")
```
