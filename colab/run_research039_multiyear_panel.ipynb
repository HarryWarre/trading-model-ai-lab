```python
import os, pathlib, subprocess
REPO = 'HarryWarre/trading-model-ai-lab'
WORK = '/content/trading-model-ai-lab'
if not pathlib.Path(WORK).exists():
    subprocess.run(['git', 'clone', f'https://github.com/{REPO}.git', WORK], check=True)
else:
    subprocess.run(['git', '-C', WORK, 'pull', '--ff-only'], check=True)
os.chdir(WORK)
subprocess.run(['pip', 'install', '-q', 'pandas', 'pytest'], check=True)
```

```python
from pathlib import Path
import subprocess

root = Path('/content/drive/MyDrive/trading-model-ai-lab')
raw = root / 'data' / 'histdata_raw'
out = root / 'data' / 'histdata_m1_5m_15_2023_2025.csv'
manifest = raw / 'histdata_multiyear_manifest.csv'
if not manifest.exists():
    raise FileNotFoundError(f'Missing download manifest: {manifest}')
cmd = [
    'python', 'colab/prepare_histdata_m1_panel.py',
    '--raw-dir', str(raw),
    '--years', '2023,2024,2025',
    '--output', str(out),
    '--min-assets', '15',
]
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
print('DONE: panel and manifest written to', out.parent)
```

```python
import json
from pathlib import Path

manifest_path = Path('/content/drive/MyDrive/trading-model-ai-lab/data/histdata_m1_5m_15_2023_2025.manifest.json')
data = json.loads(manifest_path.read_text())
print(json.dumps(data['output'], indent=2))
assert len(data['output']['assets']) == 15
assert data['output']['years_requested'] == [2023, 2024, 2025]
print('DONE: 15-asset panel validation')
```
